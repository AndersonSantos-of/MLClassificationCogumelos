`Dicionário do que significa cada valor dentros da colunas`

Esse conjunto de dados corresponde ao clássico banco de cogumelos das famílias Agaricus e Lepiota.

**classe**: *e* significa comestível e *p* significa venenoso.

**forma_chapeu** (formato do topo do cogumelo): *b* é arredondado como um sino, *c* é cônico, *f* é achatado, *k* tem uma saliência no meio, *s* é afundado no centro, e *x* é convexo, curvado para cima.

**superficie_chapeu** (textura do topo): *f* é fibrosa, *g* tem sulcos ou ranhuras, *s* é lisa, *y* é escamosa.

**cor_chapeu**: *b* é bege, *c* é canela, *e* é vermelho, *g* é cinza, *n* é marrom, *p* é rosa, *r* é verde, *u* é roxo, *w* é branco, *y* é amarelo.

**machucados**: *t* significa que o cogumelo muda de cor ou libera algum líquido quando é apertado ou cortado; *f* significa que não muda.

**odor**: *a* é cheiro de amêndoa, *c* é cheiro de creosoto, *f* é fétido, *l* é cheiro de anis, *m* é cheiro de mofo, *n* é sem cheiro, *p* é um cheiro forte e penetrante, *s* é um cheiro picante, *y* é cheiro de peixe.

**fixacao_branquia** (como as lâminas por baixo do chapéu se prendem ao talo): *a* significa presas ao talo, *f* significa soltas.

**espacamento_branquia**: *c* é próximas umas das outras, *w* é bem apertadas e numerosas.

**tamanho_branquia**: *b* é largas, *n* é estreitas.

**cor_branquia**: *b* é bege, *e* é vermelho, *g* é cinza, *h* é cor de chocolate, *k* é preto, *n* é marrom, *o* é laranja, *p* é rosa, *r* é verde, *u* é roxo, *w* é branco, *y* é amarelo.

**forma_caule** (formato do talo): *e* significa que ele alarga perto da base, *t* significa que ele afina perto da base.

**raiz_caule** (como a base do talo termina): *?* é uma informação que não foi registrada, *b* é em formato de bulbo, *c* é em formato de clava, *e* é uniforme em toda a extensão, *r* é com uma raiz visível saindo da base.

**superficie_caule_acima_anel** e **superficie_caule_abaixo_anel** (textura do talo acima e abaixo do anel): *f* é fibrosa, *k* é sedosa, *s* é lisa, *y* é escamosa.

**cor_caule_acima_anel** e **cor_caule_abaixo_anel**: *b* é bege, *c* é canela, *e* é vermelho, *g* é cinza, *n* é marrom, *o* é laranja, *p* é rosa, *w* é branco, *y* é amarelo.

**cor_veu** (cor da membrana fina que às vezes cobre partes do cogumelo quando ele ainda é jovem): *n* é marrom, *o* é laranja, *w* é branco, *y* é amarelo.

**numero_anel** (quantos anéis o talo tem): *n* é nenhum, *o* é um, *t* é dois.

**tipo_anel** (formato do anel que fica no talo): *e* é um anel que desaparece com o tempo, *f* é alargado, *l* é grande, *n* é sem anel, *p* é um anel pendurado.

**cor_impresao_esporos** (cor que fica marcada quando se coloca o chapéu sobre um papel e os esporos caem): *b* é bege, *h* é cor de chocolate, *k* é preto, *n* é marrom, *o* é laranja, *r* é verde, *u* é roxo, *w* é branco, *y* é amarelo.

**populacao** (quantos cogumelos costumam aparecer juntos no mesmo local): *a* é em grande quantidade, *c* é agrupados bem próximos, *n* é numerosos, *s* é espalhados, *v* é vários, *y* é sozinho.

**habitat** (onde o cogumelo costuma crescer): *d* é em matas ou bosques, *g* é em gramados, *l* é entre folhas caídas, *m* é em campos abertos, *p* é em trilhas ou caminhos, *u* é em áreas urbanas, *w* é em terrenos baldios.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_cogumelos = pd.read_csv('./data/cogumelos_tratados.csv')
df_cogumelos.head()
```

```python
# definindo as colunas que irão aparecer no gráfico

colunas = df_cogumelos.columns
```

countplot é um gráfico de barras usado para váriaveis categóricas

```python
# visualizando a distribuição dos valores em cada coluna

fig, axes = plt.subplots(6, 4, figsize=(18, 12))
axes = axes.flatten() # transforma a grade em uma lista de 24 objetos

for ax, cols in zip(axes, colunas): # emparelha cada espaço ax com uma coluna
    sns.countplot(data=df_cogumelos, x=cols, ax=ax)
    ax.grid(True)

for ax in axes[len(colunas):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# distribuição dos cogumelos comestíveis/venenosos para cada categoria de cada coluna
# p -> venenoso | e -> comestível

fig, axes = plt.subplots(7, 3, figsize=(12, 18))
axes = axes.flatten()

for ax, cols in zip(axes, x):
    sns.countplot(data=df_cogumelos, x=cols, ax=ax, hue='classe')
    ax.grid(True)


plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency
import numpy as np
```

```python
alvo = 'classe'

colunas_caracteristicas = ['forma_chapeu', 'superficie_chapeu', 'cor_chapeu',
       'machucados', 'odor', 'fixacao_branquia', 'espacamento_branquia',
       'tamanho_branquia', 'cor_branquia', 'forma_caule', 'raiz_caule',
       'superficie_caule_acima_anel', 'superficie_caule_abaixo_anel',
       'cor_caule_acima_anel', 'cor_caule_abaixo_anel', 'cor_veu',
       'numero_anel', 'tipo_anel', 'cor_impresao_esporos', 'populacao',
       'habitat']

resultados = []
```

```python
# teste para mediir a força da relação de cada característica com a classe

for coluna in colunas_caracteristicas:
    # cria uma tabela de cruzamento entre o alvo e uma caracteristicas
    tabela = pd.crosstab(df_cogumelos[coluna], df_cogumelos[alvo])

    # roda o teste na tabela de contagens e devolve quatro informações
    chi2, p, dof, esperado = chi2_contingency(tabela)

    # preparam os números necessários para deixar o resultado comparável entre colunas diferentes
    n = tabela.sum().sum()
    r, k = tabela.shape
    grau_liberdade_minimo = min(r - 1, k - 1)

    associacao = np.sqrt((chi2 / n) / grau_liberdade_minimo)

    resultados.append({'caracteristica': coluna, 'associacao': round(associacao, 3), 'p_valor':p})

resultado_final = pd.DataFrame(resultados).sort_values('associacao', ascending=False)
print(resultado_final.to_string(index=False))

# associacao mostra a força da relação com a classe.  
# p_valor mostra a chance de essa relação ter aparecido só por coincidência.
```
