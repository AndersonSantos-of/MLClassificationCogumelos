```python
# removendo a coluna que tem apenas um único valor único

df_cogumelos.drop(columns='veil-type', inplace=True)
```

```python
# traduzindo e renomenado o nome das colunas

dicionario_colunas = {
    'class': 'classe',
    'cap-shape': 'forma_chapeu',
    'cap-surface': 'superficie_chapeu',
    'cap-color': 'cor_chapeu',
    'bruises': 'machucados',
    'gill-attachment': 'fixacao_branquia',
    'gill-spacing': 'espacamento_branquia',
    'gill-size': 'tamanho_branquia',
    'gill-color': 'cor_branquia',
    'stalk-shape': 'forma_caule',
    'stalk-root': 'raiz_caule',
    'stalk-surface-above-ring': 'superficie_caule_acima_anel',
    'stalk-surface-below-ring': 'superficie_caule_abaixo_anel',
    'stalk-color-above-ring': 'cor_caule_acima_anel',
    'stalk-color-below-ring': 'cor_caule_abaixo_anel',
    'veil-type': 'tipo_veu',
    'veil-color': 'cor_veu',
    'ring-number': 'numero_anel',
    'ring-type': 'tipo_anel',
    'spore-print-color': 'cor_impresao_esporos',
    'population': 'populacao',
}

df_cogumelos.rename(columns=dicionario_colunas, inplace=True)
```

```python
print(f'Linhas: {df_cogumelos.shape[0]} | Colunas: {df_cogumelos.shape[1]}')
df_cogumelos.head()
```
